```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("StudentsPerformance_csv")
print("Dataset shape (rows, columns):", df.shape)
print("Column name:", df.columns.tolist())

print("First 5 rows of the dataset:")
print(df.head())

print("Summary statistics:")
print(df.describe())

print("Data type and null info")
print(df.info())

```

```python
print("Average score by columns:")

df[["GPA", "StudyTimeWeekly", "Absences"]].mean()
```

```python
print("Maximum values by columns:")

df[["GPA", "StudyTimeWeekly", "Absences"]].max()
```

```python
print("Maximum values by columns:")

df[["GPA", "StudyTimeWeekly", "Absences"]].min()
```

```python
# correlation
print("Correlation")
df[["GPA", "StudyTimeWeekly", "Absences"]].corr()
```

```python
import numpy as np

# GPA asosida foizli ball yaratish (0–100)
df["Total score"] = df["GPA"] / 4 * 100

# O'rtacha ball
df["average score"] = df["Total score"]

# O'tgan yoki o'tmaganini aniqlash
df["passed"] = np.where(df["GPA"] >= 2.0, "Yes", "No")

# Natijani ko'rish
print("New colums preview (total, average, passed):")
df[["GPA", "Total score", "average score", "passed"]].head()
```

```python
import numpy as np

# 1. Tutoring bo'yicha o'rtacha GPA
tutoring_group = df.groupby("Tutoring")[["GPA", "StudyTimeWeekly"]].mean()
print("Student performance by tutoring:")
print(tutoring_group)

# 2. Gender bo'yicha o'rtacha natijalar
gender_avg = df.groupby("Gender")[["GPA", "StudyTimeWeekly", "Absences"]].mean()
print("Average score by gender:")
print(gender_avg)

# 3. O'tgan va o'tmagan talabalar foizi
df["passed"] = np.where(df["GPA"] >= 2.0, "Yes", "No")

pass_rate = df["passed"].value_counts(normalize=True) * 100
print("Pass rate:\n", pass_rate)
```

```python
import numpy as np

# Convert data to NumPy array
scores = df[["GPA", "StudyTimeWeekly", "Absences"]].values

# Mean, Standard Deviation
print("Numpy Mean:", np.mean(scores, axis=0))
print("Numpy Std:", np.std(scores, axis=0))

# Z-score (normalization)
z_score = (scores - np.mean(scores, axis=0)) / np.std(scores, axis=0)

print("Z-scores (normalized numbers):")
print(z_score[:5])
```

```python
import numpy as np

# How many students scored above 90?
top_scores = df[df["GPA"] / 4 * 100 >= 90]
print("Top scored students above 90:", len(top_scores))

# How many failed?
df["passed"] = np.where(df["GPA"] >= 2.0, "Yes", "No")

failed = df[df["passed"] == "No"]
print("Failed from exam:", len(failed))
```
