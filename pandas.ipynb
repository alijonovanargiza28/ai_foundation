```python
data_with_missing ={
    "cars":["BMW","Volvo",6,"Tesla","Ferrari", None, "Ford"],
    "counts":[3,5,None, 5,3,None,5]
}
df_missing = pd.DataFrame(data_with_missing)
print(df_missing)
```

```python
df_missing.isnull().sum()
```

```python
df_missing.dropna(axis=1)
```

```python
df_missing.dropna(how="all")
```

```python
df_missing.dropna(subset=['cars'])
```

```python
df_missing.fillna(0)
```

```python
df_missing.fillna({"cars":"unknown", "counts":0})
```
